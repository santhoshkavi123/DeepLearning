```python
from typing import List
```

```python
class Solution:

    def frequency_generator(self, char_lst : List[str]):
        """
            Function : Generates the Frequency of individual characters from a given list
            args:
                char_lst : List[str] : contains characters of a particular word
            returns:
                freq_lst : Dict : contains frequencies of individual characters
        """
        freq_list = {} # Initialise empty dictionary
        for char in range(0, len(char_lst)):
            if char_lst[char] not in freq_list.keys():
                freq_list[char_lst[char]] = 1
            else:
                freq_list[char_lst[char]] += 1

        return freq_list


    def bucket_sort(self, nums : List[str], freq_list: dict):        
        # Empty dictionary with pre-defined shape
        bucket_sorted_list = {key: [] for key in range(0, len(nums)+1)} 

        for key, value in freq_list.items():
            bucket_sorted_list[value].append(key)

        return bucket_sorted_list

    def reverse_freq_extractor(self, bucket_sorted_list:dict, k:int):
        top_k_list = []
        for i in range(len(bucket_sorted_list) - 1, -1, -1):  # iterate from end to start
            if bucket_sorted_list[i]:  # check if bucket is not empty
                for item in bucket_sorted_list[i]:
                    if len(top_k_list) < k:
                        top_k_list.append(item)
                    else:
                        break

            if len(top_k_list) >= k:
                break
        
        return top_k_list


    def topKFrequent(self, nums: List[int], k: int) -> List[int]:
        # Use frequency_generator to freq of individual characters
        freq_list = self.frequency_generator(nums)

        # Use Bucket Sort 
        freq_list = self.bucket_sort(nums,freq_list)

        # Get Top K list
        freq_list = self.reverse_freq_extractor(freq_list, k)
        
        return freq_list
        
```

```python
func = Solution()
func.topKFrequent(nums = [1,2,2,2,3,3,3], k = 2)
```
