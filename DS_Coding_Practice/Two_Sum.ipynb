```python
from typing import List
```

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:

        hashmap_dict = {} 

        for i in range(0, len(nums)):

            diff = target - nums[i]
            if nums[i] in hashmap_dict.keys():
                return [hashmap_dict[nums[i]],i]
            else:
                hashmap_dict[diff] = i
```

```python
func = Solution()
func.twoSum(nums = [3, 4, 5, 6], target = 7)
```
