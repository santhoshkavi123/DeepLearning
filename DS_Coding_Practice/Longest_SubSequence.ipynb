```python
from typing import List
```

```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        left = 0
        right = 1
        max_substring = 0

        while right <= len(s):
            sub_string = len(set(s[left : right]))
            original_string = len(s[left : right])

            if sub_string == original_string:
                right = right + 1
                max_substring = max(max_substring, sub_string)
            else:
                left = left + 1
                right = left + 1
                max_substring = max(max_substring, sub_string)
    
        return max_substring
```
