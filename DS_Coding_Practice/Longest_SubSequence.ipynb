```python
from typing import List
```

```python
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:        
        
        if len(nums) >0:
            numsSet = set(nums) # we get unique elements in a list
            longest = 1 # Record the maximum sequence length

            for i in numsSet:
                if i-1 not in numsSet: # Check if this is the start of the sequence
                    sub_sequence_length = 1
                    while (i+1) in numsSet:
                        sub_sequence_length += 1
                        longest = max(longest, sub_sequence_length)
                        i = i+1
        else:
            longest = 0

        return longest
```

```python
func = Solution()
func.longestConsecutive(nums = [2,20,4,10,3,4,5])
```
