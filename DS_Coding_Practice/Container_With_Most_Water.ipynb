```python
from typing import List
```

```python
class Solution:
    def maxArea(self, heights: List[int]) -> int:
        max_area = 0
        left_index = 0
        right_index = len(heights)-1 

        while left_index < right_index:

            if heights[left_index] > heights[right_index]:
                new_area = (right_index - left_index) * heights[right_index]
                max_area = max_area if max_area > new_area else new_area
                right_index = right_index - 1

            else:
                new_area = (right_index - left_index) * heights[left_index]
                max_area = max_area if max_area > new_area else new_area
                left_index = left_index + 1

        return max_area
        
        
```
