```python
import heapq
from visualize_heapq import visualize_heapq

A = [-4, 3, 1, 0, 2, 4, 10, 8, 12, 9] # let's we have an array 
# we are first making an array into heapify
# by default heapq makes the array as min heapify
heapq.heapify(A)
```

```python
# visualizing the heapq
visualize_heapq(A)
```

#### Heap Push - Inserting an element into Heap

Time Complexity for heappush is O(log n)

```python
heapq.heappush(A, 4) # A is my array and the inserting the element 4


visualize_heapq(A)
```

##### Heap Pop - Extracting minimum value element of the array

Time Complexity for heappop is O(log n)

```python
minn = heapq.heappop(A)


print(f"Minimum element extracted from the Heapq : {minn}")
visualize_heapq(A)
```

#### Heap Sort - Sorting the array of elements in ascending order 

Time Complexity for Heap Sort is O(n log n) and Space Complexity can be O(1)

```python
arr = []
while A:
    minn = heapq.heappop(A)
    arr.append(minn)
```

```python
arr
```

##### Heap Push Pop : Basically it will push and pop from the list 

Time Complexity for Push and Pop is O(log n)

```python
heapq.heappushpop(A, 99)
```

```python
visualize_heapq(A)
```

we can clearly see that first it will pop and then push the element

#### Max Heap

We don't have max heap in the heap.heapify operation

```python
B = [-4, 3, 1, 0, 2, 5, 10, 8, 12, 9]
n = len(B)

# We are just negating the values and then picking the max value
for i in range(n):
    B[i] = -B[i]

heapq.heapify(B)

B

visualize_heapq(B)

# After we negate the value we have the largest value as the minimum value of the heap
```
