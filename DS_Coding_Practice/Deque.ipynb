```python
from typing import Optional
```

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right

class Solution:   
    def isSubtree(self, root: Optional[TreeNode], subRoot: Optional[TreeNode]) -> bool:

        if not root or not subRoot:
            return False
        
        # Find the node in root that matches subRoot root value
        queue = deque([root])
        target_node = None
        
        while queue:
            node = queue.popleft()
            if node.val == subRoot.val:
                target_node = node
                break
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        
        if not target_node:
            return False

        # Now compare both trees using BFS
        queue = deque([target_node])
        st_queue = deque([subRoot])
        
        while queue and st_queue:
            node = queue.popleft()
            node_sub = st_queue.popleft()

            if node.val != node_sub.val:
                return False

            # Structural check
            if (node.left and not node_sub.left) or (not node.left and node_sub.left):
                return False
            if (node.right and not node_sub.right) or (not node.right and node_sub.right):
                return False

            # Add children if both exist
            if node.left and node_sub.left:
                queue.append(node.left)
                st_queue.append(node_sub.left)
            if node.right and node_sub.right:
                queue.append(node.right)
                st_queue.append(node_sub.right)

        # Both should end together
        return not queue and not st_queue
```
