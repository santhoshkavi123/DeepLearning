```python
class Solution:
    def isValid(self, s: str) -> bool:
        default_dict = {"(":")", "{":"}", "[":"]"}
        open_position_index, i = 0, 0
        open_position_symbol = []

        # Case where the s is only single character or None
        if len(s) <= 1:
            return False

        while i < len(s):
            if (i == 0) and (s[i] not in default_dict.keys()):
                return False 
            else:
                if s[i] in default_dict.keys():
                    open_position_index += 1
                    open_position_symbol.append(s[i])
                    i+=1
                else:
                    if open_position_index != 0:
                        if default_dict[open_position_symbol[-1]] == s[i]:
                            open_position_index -=1
                            open_position_symbol.pop(-1)
                            i+=1
                        else:
                            return False
                    else:
                        return False
                    
        if open_position_index == 0:
            return True
        else:
            return False

        return True
```

```python
func = Solution()
func.isValid(s = "[(])")
```
