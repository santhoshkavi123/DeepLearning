```python
from typing import List
```

```python
class Solution:

    def frequency_generator(self, char_lst : List[str]):
        """
            Function : Generates the Frequency of individual characters from a given list
            args:
                char_lst : List[str] : contains characters of a particular word
            returns:
                freq_lst : Dict : contains frequencies of individual characters
        """
        freq_list = {} # Initialise empty dictionary
        for char in range(0, len(char_lst)):
            if char_lst[char] not in freq_list.keys():
                freq_list[char_lst[char]] = 1
            else:
                freq_list[char_lst[char]] += 1

        return freq_list

    def check_anagram(self, freq_lst1 : List[str], freq_lst2 : List[str]):
        """
            Function : Check whether the given two word list qualifies to be said as anagram
            args:
                freq_lst1 : List[str] : contains frequencies of individual characters
                freq_lst2 : List[str] : contains frequencies of individual characters
            returns:
                bool : if two frequency list are same returns true else false
        """
        for key in freq_lst1.keys():

            # Checks if the key is present in other list
            if key not in freq_lst2.keys():
                return False

            # Checks if the value is present in other list is same or not
            if freq_lst1[key] != freq_lst2[key]:
                return False
        return True


    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        """
            Function : Using HashMap by creating a unique key we group similar entires
            args: 
                strs: contains the words
            returns:
                final_list : Contains pairs of word that having unique key
        """
        final_lst = {}
        for word_index in range(0, len(strs)): # Atleast there should be one word
            current_word = list(strs[word_index]) # get list of strings
            current_word_freq_lst = self.frequency_generator(current_word) # gets frequency_list   

            # Creating a Unique Key
            sorted_by_key= dict(sorted(current_word_freq_lst.items()))
            sorted_current_word = ''.join(sorted_by_key.keys()) \
                                    + ''.join(str(sorted_by_key.values())) #sorted characters and joined
            
            if sorted_current_word not in final_lst.keys():
                final_lst[sorted_current_word] = [strs[word_index]]
            else:
                final_lst[sorted_current_word].append(strs[word_index])

        return list(final_lst.values())
```

```python
func = Solution()
func.groupAnagrams(strs = ["act","pots","tops","cat","stop","hat"])
```
