```python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        s = list(s)
        t = list(t)

        if len(s) == len(t):
            char1_dict = {} # Maintains how many times a char is present in s
            char2_dict = {} # Maintains how many times a char is present in t

            for i in range(0, len(s)):

                # Char1_Dict
                if s[i] not in char1_dict.keys():
                    char1_dict[s[i]] = 1
                else:
                    char1_dict[s[i]] = char1_dict[s[i]] +1

                # Char2_Dict
                if t[i] not in char2_dict.keys():
                    char2_dict[t[i]] = 1
                else:
                    char2_dict[t[i]] = char2_dict[t[i]] +1

            # Compare both the dictionaries
            for key, value in char1_dict.items():

                if key in char2_dict.keys():        
                    if char1_dict[key] == char2_dict[key]:
                        pass
                    else:
                        return False
                else:
                    return False

            return True
        
        else:
            return False
```

```python
a = Solution()
a.isAnagram(s = "xx", t= "x")
```

# Solution 2

s = "racecar"
t = "carrace"

if we sort the 's' we get  `aaccerr`
similarly, if we sort the 't' we get `aaccerr`

sorted is a built-in function in Python that sorts the characters of a string.

```python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        return sorted(s) == sorted(t)
```

```python
a = Solution()
a.isAnagram(s = "racecar", t= "carrace")
```
