```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next



class Solution:
    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:
        curr = head
        dummy = ListNode(0, head)
        prev = dummy
        len_of_listed_node = 1
        curr_index = 0

        # To measure the length of linkedlist
        while curr.next:
            len_of_listed_node += 1
            curr = curr.next

        target_to_remove = len_of_listed_node - n

        curr = head
        # Delete Operation
        while curr:
            if curr_index == target_to_remove:
                prev.next = curr.next

            prev, curr = curr, curr.next
            curr_index += 1

        return dummy.next

        
```
