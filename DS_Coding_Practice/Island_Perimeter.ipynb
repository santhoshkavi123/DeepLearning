```python
from typing import List
import collections
```

```python
class Solution:
    def islandPerimeter(self, grid: List[List[int]]) -> int:
        # Grid is None
        if not grid:
            return 0
        
        perimeter = 0
        rows, cols = len(grid), len(grid[0])
        
        visit = set() # Node that are visited are recorded in this position
        directions = [[0, -1], [0, 1], [1, 0], [-1, 0]] # These are the possible directions that we can move


        def bfs(row, col):
            nonlocal perimeter
            q = collections.deque()
            q.append((row, col))

            while q:
                row, col = q.popleft()
                visit.add((row, col))
                print("Master:", row, col)
                for dr, dc in directions:
                    print("Directions", dr, dc)
                    row_dr, col_dr = row + dr , col + dc
            
                    if (row_dr in range(rows)) and (col_dr in range(cols)) \
                                and (grid[row_dr][col_dr] == 1):
                        if ((row_dr, col_dr) not in visit):
                            q.append((row_dr, col_dr))
                    else:
                        perimeter += 1
                        print("Child", row_dr, col_dr, perimeter)

                
        for i in range(rows):
            for j in range(cols):
                if grid[i][j] == 1 and (i, j) not in visit:
                    bfs(i, j)

        
        return perimeter

```
