# LinkedList - Theory

A Linked List is like a chain of boxes:

- Each box (called a node) stores:
  - A value (val)
  - A link (pointer) to the next box in the chain (next)
- The chain starts at the head node.
- The last node points to None (meaning “end of list”).

📦➡📦➡📦➡🚫

Example: 1 -> 2 -> 3 -> None

Why use a Linked List instead of a Python List?

Advantages:
- Easy to insert/remove elements without shifting everything (unlike arrays/lists)
- Can grow without reallocating memory

Disadvantages:
- Access is sequential (you can’t jump to index k instantly)
- Uses more memory (extra pointer storage)

# LinkedList -- NodeStructure

```python
class ListNode:
    def __init__(self, val=0, next = None):
        self.val = val
        self.next = next


node1 = ListNode(1) #Contains node1 with value 1
node2 = ListNode(2) #Contains node2 with value 2
node3 = ListNode(3) #Contains node3 with value 3

node1.next = node2
node2.next = node3

head = node1 # Head points the start of the LinkedList
```

```python
node1.val, node1.next.val, node2.next.val, node3.next
```

#### 1. Traversing through the LinkedList

```python
curr = head
while curr:
    print(curr.val)
    curr = curr.next
```
