```python
class Solution:
    def isPalindrome(self, s: str) -> bool:
        # Create a string that contains only AlphaNumeric Characters
        str_alphanum = ""
        for sub_str in range(0, len(s)):
            if s[sub_str].isalnum():
                str_alphanum += s[sub_str].lower()

        # Check if the string is palindrome or not
        for str1, str2 in zip(range(0, len(str_alphanum)), \
                                range(len(str_alphanum)-1, -1, -1)):

            if str_alphanum[str1] != str_alphanum[str2]:
                return False

        
        return True
```

```python
func = Solution()
func.isPalindrome(s = "Was it a car or a cat I saw?")
```
