```python
from typing import List
```

```python
class Solution:

    def encode(self, strs: List[str]) -> str:
        encoded_string = ""
        for i in range(0, len(strs)):
            prefix_encoded_list = str(len(strs[i])) + "#"
            encoded_string += (prefix_encoded_list + strs[i])

        return encoded_string

    def decode(self, s: str) -> List[str]:
        decoded_lst = []
        i = 0
        while i < len(s):
            # Extract the length prefix
            j = i
            while s[j] != "#":
                j += 1
            length = int(s[i:j])
            decoded_str = s[j + 1 : j + 1 + length]
            decoded_lst.append(decoded_str)
            i = j + 1 + length
            
        return decoded_lst
            
```

```python
func = Solution()
func.encode(strs=["we","say",":","yes","!@#$%^&*()"])
func.decode(s = '2#we3#say1#:3#yes10#!@#$%^&*()')
```
