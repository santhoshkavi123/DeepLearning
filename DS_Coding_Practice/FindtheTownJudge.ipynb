```python
from typing import List
from collections import defaultdict
```

```python
class Solution:
    def findJudge(self, n: int, trust: List[List[int]]) -> int:
        if n==1:
            return 1 

        outdegree = defaultdict(int)
        indegree = defaultdict(int)

        for vertex, edj_vertex in trust:
            if vertex not in outdegree.keys():
                outdegree[vertex] = 1
            else:
                outdegree[vertex] += 1

        for vertex, edj_vertex in trust:
            if (edj_vertex not in indegree.keys()):
                if (edj_vertex not in outdegree.keys()): 
                    indegree[edj_vertex] = 1
            else:
                indegree[edj_vertex] += 1

        len_of_vertex = len(outdegree.keys()) + len(indegree.keys()) - 1
        print(indegree, outdegree)

        for key, _ in indegree.items():
            if indegree[key] == len_of_vertex:
                return key

        return -1
        
```
